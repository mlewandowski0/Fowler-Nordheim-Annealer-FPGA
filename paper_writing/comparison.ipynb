```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Data
names = ["Intel Core Ultra 9 185H", "SpiNNaker 2", "This Work", "Bifurcation Machine"]
times = [504, 2460, 20, 0.06]
power_consumption = [11.63, 0.562, 1.42, 0]

df = pd.DataFrame({
    "Name": names,
    "Time (s)": times,
    "Power (W)": power_consumption
})

# Set seaborn theme
sns.set_theme(style="whitegrid", context="talk")

# Custom color palette: highlight "This Work"
highlight_color = 'orange'
default_color   = 'steelblue'
colors = [highlight_color if name == "This Work" else default_color for name in df["Name"]]

# Plot
fig, ax1 = plt.subplots(figsize=(12, 7))

# Barplot with custom colors
sns.barplot(
    data=df,
    x="Name",
    y="Time (s)",
    ax=ax1,
    palette=colors
)

# Annotate bars
for i, val in enumerate(df["Time (s)"]):
    ax1.text(i, val, f'~ {val}s', 
             ha='center', va='bottom', fontsize=16, fontweight='bold')

# Log scale
ax1.set_yscale('log')

# Make y-axis label bigger
ax1.set_ylabel("Time (s, log scale)", fontsize=24, labelpad=15)

# Make y-tick labels bigger
ax1.tick_params(axis='y', labelsize=20)
ax1.tick_params(axis='x', labelsize=20)

# Remove x-axis label and style the title
ax1.set_xlabel("", fontsize=16)
ax1.set_title(
    "NeuroSA: time to make 100 mln flips | Bifurcation Machine: time to solution",
    fontsize=20, fontweight='bold', pad=20
)

# Rotate and size x-ticks
#ax1.set_xticklabels(df["Name"], rotation=25, ha='right', fontsize=16)

plt.tight_layout()
plt.savefig("comparison.png", dpi=450)
plt.show()
```
