```python
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os 

class CONFIG:
    LOG_FOLDER = os.path.join(".", "logs")
    PROBLEM = "G15"

```

```python
import os
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def get_problem_sol():
    if CONFIG.PROBLEM == "G15":
        return 3050
    else:
        raise ValueError("Unknown solution to the selected problem")

def plot_group(prefixes, dir = CONFIG.LOG_FOLDER):
    paths = os.listdir(dir)
    
    if isinstance(prefixes, str):
        prefixes = [prefixes]

    x_min, x_max = float('inf'), float('-inf')

    fig = make_subplots()

    for path in paths:
        try:
            for prefix in prefixes:
                if path.startswith(prefix):
                    log_path = os.path.join(CONFIG.LOG_FOLDER, path)
                    data = pd.read_csv(log_path, sep=",", index_col=False)

                    fig.add_trace(go.Scatter(x=data["iter"], y=data["best_maxcuts"], mode='lines', name=path))

                    x_min = min(x_min, data["iter"].values[0])
                    x_max = max(x_max, data["iter"].values[-1])
                    break
        except Exception as e:
            print(f"Error: {e} for {path}")

    fig.add_trace(go.Scatter(x=[x_min, x_max], y=[get_problem_sol(), get_problem_sol()], mode='lines', 
                             name="SOTA (best known)", line=dict(dash='dash', color='black')))

    fig.update_layout(title='Interactive Plot', xaxis_title='Iteration', yaxis_title='Best Maxcuts',
                      width=1600, height=800)  # Adjust the width and height as needed
    fig.show()

plot_group(prefixes="unmodified")
```
