```python
import numpy as np
import matplotlib.pyplot as plt 
import os
import pandas as pd  
import plotly.express as px
```

```python
data_root_path = os.path.join(".", "logs")

# Function to plot using plotly
def plot_problem(data_root, problem_name="G15"):
    
    # Update data root to include "logs" directory
    data_root = os.path.join(data_root, "logs")
    
    # Loop through files in the directory
    for file in os.listdir(data_root):
        file_fullpath = os.path.join(data_root, file)
        
        # Check if it is a file
        if os.path.isfile(file_fullpath) and problem_name in file_fullpath:
            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_fullpath)
            
            # Use plotly to plot the data
            fig = px.line(df, x="iter", y="best_maxcuts", title=f"Plot for {file}")
            fig.show()

# Call the function to plot the data
plot_problem(data_root_path)
```

```python
import plotly.graph_objects as go


# Function to plot all files on a single plot using plotly
def plot_problem_on_single(data_root, problem_name="G15"):
    
    # Update data root to include "logs" directory
    data_root = os.path.join(data_root, "logs")
    
    # Create a plotly figure
    fig = go.Figure()

    # Loop through files in the directory
    for file in os.listdir(data_root):
        file_fullpath = os.path.join(data_root, file)
        
        # Check if it is a file
        if os.path.isfile(file_fullpath) and problem_name in file_fullpath:
            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_fullpath)
            
            # Add each line from the file to the figure
            fig.add_trace(go.Scatter(x=df["iter"], y=df["best_maxcuts"],
                                     mode='lines', name=file[33:-4]))
    
    # Update layout for the figure
    fig.update_layout(
        title=f"Maxcuts per Iteration for all files in {problem_name}",
        xaxis_title="Iteration",
        yaxis_title="Best Maxcuts",
        showlegend=True
    )

    # Show the plot
    fig.show()

# Call the function to plot the data
plot_problem_on_single(data_root_path)
```
